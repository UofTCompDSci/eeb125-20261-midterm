### Part 3, Question 1 (4 points)

Run the code cell below to define a function that we'll use to compute the arithmetic mean.

```python
# RUN THIS CELL, but do NOT modify the code.

def mean(pop):
    total = 0.0
    for num in pop:
        total += num
    average = total/float(len(pop))    
    return average
```

Next, we will explore a dataset documenting the species of all 55 trees found within a small patch of forest in Southern Ontario. For easier interpretation, we wil refer to each species using its common name rather than its scientific name.

To start, please execute the code provided below:

```python
# RUN THIS CELL, but do NOT modify the code.

trees_file = open("trees.csv", "r")
trees_lines = trees_file.readlines()
trees_header = trees_lines[0]
trees_data = trees_lines[1:]

all_species = []

for line in trees_data:
    line_dat = line.strip().split(",")
    species = line_dat[2]
    all_species.append(species)

species_proportions = {}
for species in set(all_species):
    species_proportions[species] = all_species.count(species) / len(all_species)
```
